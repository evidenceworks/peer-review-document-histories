# Public peer review: Figures

Rebuild all five article figures from published data and execute the independent 2,008-check numerical audit. This is not the original source-capture replay.


```python
import os
from pathlib import Path
import subprocess
import sys

repository = os.environ.get("AOIR_REPO_ROOT")
if repository:
    root = Path(repository).expanduser().resolve()
else:
    root = Path("peer-review-document-histories").resolve()
    if not (root / "README.md").is_file():
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/evidenceworks/peer-review-document-histories.git", str(root)], check=True)
assert (root / "analysis" / "result_manifest.json").is_file()
print("Research files ready:", root)
```

```python
import shutil
import subprocess
import tempfile
from pathlib import Path

font_family = (subprocess.run(["fc-match", "-f", "%{family}", "Nimbus Roman"], capture_output=True, text=True).stdout if shutil.which("fc-match") else "")
if "Nimbus Roman" not in font_family:
    subprocess.run(["apt-get", "update", "-qq"], check=True)
    subprocess.run(["apt-get", "install", "-y", "-qq", "fonts-urw-base35"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(root / "analysis" / "FigureProduction" / "Code" / "requirements.txt")], check=True)
output = Path(tempfile.mkdtemp(prefix="peer-review-figures-")) / "five_figures"
subprocess.run([sys.executable, "-B", str(root / "scripts" / "figure_check.py"), "--output", str(output)], check=True)
print("Verified figure files:", output)
```
