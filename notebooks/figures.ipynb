# Figures

Rebuild the five study figures from the released data and check their numerical values.


```python
import os
from pathlib import Path
import subprocess
import sys

repository = os.environ.get("AOIR_REPO_ROOT")
if repository:
    root = Path(repository).expanduser().resolve()
else:
    root = Path("peer-review-document-histories").resolve()
    if not (root / "README.md").is_file():
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/evidenceworks/peer-review-document-histories.git", str(root)], check=True)
assert (root / "analysis" / "result_manifest.json").is_file()
print("Research files ready:", root)
```

```python
import subprocess
import sys
import tempfile
from pathlib import Path

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(root / "analysis" / "FigureProduction" / "Code" / "requirements.txt")], check=True)
output = Path(tempfile.mkdtemp(prefix="review-figures-")) / "figures"
subprocess.run([sys.executable, "-B", str(root / "scripts" / "figure_check.py"), "--output", str(output)], check=True)
print("Figures saved to:", output)
```
