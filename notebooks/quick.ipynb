# Quick check

Recalculate the principal review-event and document-history results from the released data.


```python
import os
from pathlib import Path
import subprocess
import sys

location = os.environ.get("AOIR_REPO_ROOT")
root = Path(location).resolve() if location else (Path.cwd() / "peer-review-document-histories").resolve()
if not (root / "scripts" / "reproduce_results.py").is_file():
    subprocess.run(["git", "clone", "--depth", "1",
                    "https://github.com/evidenceworks/peer-review-document-histories.git",
                    str(root)], check=True)
print("Repository:", root.name)
```

```python
archive = Path.cwd() / "outputs" / "aOIR_Quick.zip"
subprocess.run([sys.executable, "-B",
                str(root / "scripts" / "quick_check.py"),
                "--output", str(archive)], check=True)
```
